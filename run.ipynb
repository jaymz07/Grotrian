```python
import matplotlib.pyplot as plt
from grotrian import Grotrian
```

```python
e = Grotrian('input_files/example.yaml')

e.draw()
```

```python
e = Grotrian('input_files/hydrogen.yaml')

e.draw(delta_c=0.5, delta_p=0.4)
```

```python
e = Grotrian('input_files/rubidium.yaml')
e.draw()
```

```python
e = Grotrian('input_files/rubidium-87.yaml')

e.draw(text_offset=0.01, delta_c=0.5, delta_p=0.3, ax_expan_r=0.25)

# Remove y-axis
plt.gca().get_yaxis().set_visible(False)

#plt.savefig('example_outputs/rubidium-87.png', bbox_inches='tight', dpi=300)
```

```python
e = Grotrian('input_files/rubidium-85.yaml')

e.draw(delta_c=0.5, delta_p=0.3, ax_expan_r=0.25)

# Remove y-axis
plt.gca().get_yaxis().set_visible(False)

#plt.savefig('example_outputs/rubidium-85.png', bbox_inches='tight', dpi=300)
```

```python
r87 = Grotrian('input_files/rubidium-87.yaml')
r85 = Grotrian('input_files/rubidium-85.yaml')

fig, (ax1, ax2) = plt.subplots(1,2, sharey=True, figsize=(10,5))

r85.draw(ax=ax1, delta_c=0.5, delta_p=0.3, ax_expan_r=0.25)
r87.draw(ax=ax2, delta_c=0.5, delta_p=0.3, ax_expan_r=0.25)

#plt.savefig('example_outputs/rubidium_combined.png', bbox_inches='tight', dpi=300)
```
